## Part A – Machine Initialization and Constants

```python
import random

# Number of machines
M = 3
# Machine names
machines = ['A', 'B', 'C']
# Number of plays
N_play = 10
# Exploration probability
epsilon = 0.2

print(f"Number of machines (M): {M}")
print(f"Machine names: {machines}")
print(f"Number of plays (N_play): {N_play}")
print(f"Exploration probability (ε): {epsilon}")
```

## Part B – Create knowledge, means update N, R and Q based on initial test

Initialize the `N` (number of times played), `R` (total reward), and `Q` (estimated value) tables for each machine. These will be dictionaries where keys are machine names.

```python
# Initialize N, R, and Q tables for each machine
N = {machine: 0 for machine in machines}  # N(a): number of times machine 'a' has been selected
R = {machine: 0 for machine in machines}  # R(a): total reward obtained from machine 'a'
Q = {machine: 0.0 for machine in machines} # Q(a): current estimated value of machine 'a'

print("Initial N (play count):", N)
print("Initial R (total reward):", R)
print("Initial Q (estimated value):", Q)
```

## Part C – ε-Greedy Decision, Part D – Reward Collection, and Part E – Q-Value Update

This section implements the core ε-greedy strategy loop. For each play, it decides whether to explore or exploit, collects the user-provided reward, and updates the `N`, `R`, and `Q` values.

```python
all_rewards = [] # To keep track of all rewards for total calculation

print("\n--- Starting Multi-Armed Bandit Simulation ---")

for play in range(1, N_play + 1):
    print(f"\n--- Play {play}/{N_play} ---")

    # 1. Generate a random number r between 0 and 1.
    r = random.random()

    selected_machine = None

    # 2. If r < ε, perform exploration:
    if r < epsilon:
        selected_machine = random.choice(machines)
        print(f"Exploration: Randomly selected machine {selected_machine}")
    # 3. Otherwise, perform exploitation:
    else:
        # Select the machine having the highest current Q-value.
        # Handle cases where multiple machines have the same highest Q-value
        max_q_value = -float('inf')
        potential_machines = []
        for machine, q_value_item in Q.items(): # Renamed 'q_value' to 'q_value_item' to avoid conflict later
            if q_value_item > max_q_value:
                max_q_value = q_value_item
                potential_machines = [machine]
            elif q_value_item == max_q_value:
                potential_machines.append(machine)

        selected_machine = random.choice(potential_machines) # Break ties randomly
        print(f"Exploitation: Selected machine {selected_machine} based on Q-value ({Q[selected_machine]:.2f})")

    # 4. Ask the user to enter the reward obtained from the selected machine.
    while True:
        try:
            reward_input = input(f"Enter reward for machine {selected_machine}: ")
            current_reward = int(reward_input)
            break
        except ValueError:
            print("Invalid input. Please enter an integer for the reward.")

    all_rewards.append(current_reward)

    # 5. Update the number of times the selected machine has been played.
    N[selected_machine] += 1

    # Update total reward for the selected machine
    R[selected_machine] += current_reward

    # 6. Update its Q-value using the incremental sample-average formula
    # Q_new(a) = Q(a) + (R - Q(a)) / N(a)
    Q[selected_machine] = Q[selected_machine] + (current_reward - Q[selected_machine]) / N[selected_machine]

    print(f"  Updated N: {N}")
    print(f"  Updated R: {R}")

    # Correct way to format Q-values for display without nested f-string issues
    formatted_Q_for_display = {machine: f'{q_val:.2f}' for machine, q_val in Q.items()}
    print(f"  Updated Q: {formatted_Q_for_display}")
```

## Part F – Display Results and Total Reward

Display the final participation count (`N`), total rewards (`R`), and estimated Q-values (`Q`) for all machines. Also, calculate the total reward obtained after all plays.

```python
import pandas as pd

print("\n--- Simulation Complete ---")

# Create a DataFrame for better display of results
results_df = pd.DataFrame({
    'Machine': machines,
    'Plays (N)': [N[m] for m in machines],
    'Total Reward (R)': [R[m] for m in machines],
    'Estimated Q-Value (Q)': [Q[m] for m in machines]
})

print("\nParticipation Count, Total Rewards, and Estimated Q-Values:")
display(results_df)

# Calculate total reward the player received after 10th play
total_reward_player = sum(all_rewards)
print(f"\nTotal reward the player received after {N_play} plays: {total_reward_player}")
```
