# Upper Confidence Bound (UCB) Multi-Armed Bandit Implementation

This notebook implements a 3-armed bandit problem where machine selection is guided by the UCB algorithm. We will use the incremental sample-average formula to update Q-values.

```python
import numpy as np
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
c = 2 # Exploration parameter

# Reward Patterns as provided
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Initialize tables
N = {m: 0 for m in machines}   # Participation count
R_total = {m: 0 for m in machines} # Total rewards
Q = {m: 0.0 for m in machines} # Estimated values
UCB = {m: float('inf') for m in machines} # UCB values

print(f"Starting UCB for {N_play} plays with automated rewards.\n")

# Main Loop
for play in range(1, N_play + 1):
    print(f"--- Play {play} ---")

    # Part C - Calculate UCB and choose machine with highest UCB
    for m in machines:
        if N[m] == 0:
            UCB[m] = float('inf')
        else:
            exploration_term = c * math.sqrt(math.log(play) / N[m])
            UCB[m] = Q[m] + exploration_term

    # Select machine with highest UCB
    selected_machine = max(UCB, key=UCB.get)
    print(f"Play with machine {selected_machine}")

    # Part D - Reward Collection (Automated)
    # Get the reward based on how many times this machine has been played
    pattern = reward_patterns[selected_machine]
    reward_index = N[selected_machine] % len(pattern)
    reward = pattern[reward_index]
    print(f"Reward obtained: {reward}")

    # Part E - Q-Value Update
    N[selected_machine] += 1
    # Incremental sample-average formula: Q_n = Q_n-1 + 1/n * (R - Q_n-1)
    error = reward - Q[selected_machine]
    Q[selected_machine] = Q[selected_machine] + (1.0 / N[selected_machine]) * error
    R_total[selected_machine] += reward

    # Display current state
    print(f"Updated Q({selected_machine}): {Q[selected_machine]:.2f}")
    print(f"Participation N: {N}")
    print(f"Current UCBs: { {k: round(v, 2) if v != float('inf') else 'inf' for k, v in UCB.items()} }")
    print("\n")

# Part F - Display Results
print("=== Final Results ===")
print(f"Participation Count (N): {N}")
print(f"Total Rewards per machine (R): {R_total}")
print(f"Estimated Values (Q): {Q}")

total_sum_reward = sum(R_total.values())
print(f"\nTotal Reward the player received: {total_sum_reward}")
```
