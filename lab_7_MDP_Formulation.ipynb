## Assignment 1: Calculate Undiscounted and Discounted Returns

```python
# ============================================
# ASSIGNMENT 1
# Calculate Undiscounted and Discounted Returns
# ============================================

# Get number of time steps
n = int(input("Enter number of time steps/episodes: "))

# Get rewards
rewards = []

print("\nEnter reward for each time step:")

for i in range(n):
    reward = float(input(f"Reward at time step {i + 1}: "))
    rewards.append(reward)

# Get discount factor
gamma = float(input("\nEnter discount factor (gamma, between 0 and 1): "))

# --------------------------------------------
# Calculate Undiscounted Return
# --------------------------------------------

undiscounted_return = sum(rewards)

# --------------------------------------------
# Calculate Discounted Return
# G = R1 + gamma*R2 + gamma^2*R3 + ...
# --------------------------------------------

discounted_return = 0

for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]

# --------------------------------------------
# Display Results
# --------------------------------------------

print("\n========================================")
print("        RETURN CALCULATION")
print("========================================")

print("Entered Reward Values:", rewards)

print("\nUndiscounted Return:")
print(undiscounted_return)

print("\nDiscounted Return:")
print(discounted_return)

print("\nDiscount factor (gamma):", gamma)

print("========================================")
```

```python
%%writefile grid_env_det.py

# ============================================
# 2 x 3 GRID WORLD
# Deterministic MDP Environment
# ============================================

class GridWorld:

    def __init__(self):

        # ------------------------------------
        # State Space
        # ------------------------------------

        self.states = {
            "S0": (0, 0),
            "S1": (0, 1),
            "S3": (1, 1),
            "S4": (1, 2),
            "G":  (0, 2)
        }

        # ------------------------------------
        # Action Space
        # ------------------------------------

        self.actions = [
            "up",
            "down",
            "left",
            "right"
        ]

        # ------------------------------------
        # Special states
        # ------------------------------------

        self.goal_state = "G"
        self.obstacle_position = (1, 0)

        # ------------------------------------
        # Rewards
        # ------------------------------------

        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        # ------------------------------------
        # Discount factor
        # ------------------------------------

        self.gamma = 0.9

    # ========================================
    # Define Reward Function
    # ========================================

    def reward_function(self, state, action):

        current_position = self.states[state]

        # Find position produced by action
        next_position = self.get_next_position(
            current_position,
            action
        )

        # Goal
        if next_position == self.states["G"]:
            return self.goal_reward

        # Obstacle
        if next_position == self.obstacle_position:
            return self.obstacle_reward

        # Normal movement
        return self.normal_reward

    # ========================================
    # Calculate next grid position
    # ========================================

    def get_next_position(self, position, action):

        row, col = position

        if action == "up":
            row -= 1

        elif action == "down":
            row += 1

        elif action == "left":
            col -= 1

        elif action == "right":
            col += 1

        # ------------------------------------
        # Grid boundaries
        # ------------------------------------

        if row < 0 or row > 1 or col < 0 or col > 2:
            return position

        return (row, col)

    # ========================================
    # Transition Function
    # ========================================

    def transition_function(self, state, action):

        position = self.states[state]

        next_position = self.get_next_position(
            position,
            action
        )

        # ------------------------------------
        # Goal
        # ------------------------------------

        if next_position == self.states["G"]:
            return "G"

        # ------------------------------------
        # Obstacle
        # Agent remains in current state
        # ------------------------------------

        if next_position == self.obstacle_position:
            return state

        # ------------------------------------
        # Find state corresponding to position
        # ------------------------------------

        for state_name, state_position in self.states.items():

            if state_position == next_position:
                return state_name

        # If no valid transition is found
        return state

    # ========================================
    # Estimate next state
    # ========================================

    def estimate_next_state(self, current_state, action):

        return self.transition_function(
            current_state,
            action
        )

    # ========================================
    # Display MDP
    # ========================================

    def display_mdp(self):

        print("\n======================================")
        print("       2 x 3 GRID WORLD MDP")
        print("======================================")

        print("\nState Space:")
        print(self.states.keys())

        print("\nAction Space:")
        print(self.actions)

        print("\nRewards:")
        print("Normal movement :", self.normal_reward)
        print("Goal            :", self.goal_reward)
        print("Obstacle        :", self.obstacle_reward)

        print("\nDiscount Factor:")
        print(self.gamma)

    # ========================================
    # Display Transition Table
    # ========================================

    def display_transition_table(self):

        print("\n======================================")
        print("       TRANSITION FUNCTION")
        print("======================================")

        print(
            f"{'State':<10}"
            f"{'Action':<10}"
            f"{'Next State':<15}"
            f"{'Reward':<10}"
        )

        print("-" * 45)

        for state in self.states:

            for action in self.actions:

                next_state = self.transition_function(
                    state,
                    action
                )

                reward = self.reward_function(
                    state,
                    action
                )

                print(
                    f"{state:<10}"
                    f"{action:<10}"
                    f"{next_state:<15}"
                    f"{reward:<10}"
                )
```

```python
# ============================================
# MAIN PROGRAM
# ============================================

from grid_env_det import GridWorld


# Create environment
env = GridWorld()


# ============================================
# 1. Define State Space
# ============================================

print("\nSTATE SPACE")
print("-----------")

for state, position in env.states.items():
    print(f"{state} = {position}")


# ============================================
# 2. Define Action Space
# ============================================

print("\nACTION SPACE")
print("------------")

print(env.actions)


# ============================================
# 3. Define Reward Function
# ============================================

print("\nREWARD FUNCTION")
print("---------------")

print("Normal movement reward :", env.normal_reward)
print("Goal reward            :", env.goal_reward)
print("Obstacle reward        :", env.obstacle_reward)


# ============================================
# 4. Define Discount Factor
# ============================================

print("\nDISCOUNT FACTOR")
print("----------------")

print("Gamma =", env.gamma)


# ============================================
# 5. Define Transition Function
# ============================================

print("\nTRANSITION FUNCTION")
print("-------------------")

env.display_transition_table()


# ============================================
# 6. Estimate Next State
# ============================================

print("\n======================================")
print("       NEXT STATE ESTIMATION")
print("======================================")


current_state = input(
    "\nEnter current state (S0, S1, S3, S4, G): "
)

action = input(
    "Enter action (up, down, left, right): "
).lower()


# Validate state
if current_state not in env.states:

    print("\nInvalid state!")

# Validate action
elif action not in env.actions:

    print("\nInvalid action!")

else:

    next_state = env.estimate_next_state(
        current_state,
        action
    )

    reward = env.reward_function(
        current_state,
        action
    )

    print("\nCurrent State :", current_state)
    print("Action        :", action)
    print("Next State    :", next_state)
    print("Reward        :", reward)

    if next_state == "G":
        print("\nGoal reached! Episode terminates.")

    elif reward == -5:
        print("\nObstacle encountered! Episode terminates.")

    else:
        print("\nNormal transition.")
```
