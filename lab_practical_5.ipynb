## 1. Setup and Agent Definition

```python
import gymnasium as gym
import random
import numpy as np

# Define the RandomAgent class as per the assignment description
class RandomAgent:
    def __init__(self, env_name):
        self.env = gym.make(env_name)
        self.env_name = env_name
        print(f"Environment '{env_name}' created successfully.")

    def run_episode(self, max_steps=200):
        # 2. Reset the environment and 3. Observe the initial state.
        state, info = self.env.reset()
        total_reward = 0
        terminated = False
        truncated = False
        steps = 0

        print(f"\n--- Running an episode for {self.env_name} ---")
        while not terminated and not truncated and steps < max_steps:
            # 4. Identify the available actions using the environment's action space.
            # 5. Select an action randomly.
            action = self.env.action_space.sample()

            # 6. Execute the action in the environment.
            # 7. Observe the next state and reward.
            next_state, reward, terminated, truncated, info = self.env.step(action)

            total_reward += reward
            state = next_state
            steps += 1

        print(f"Episode finished after {steps} steps.")
        print(f"Total reward for this episode: {total_reward}")
        return total_reward

    def evaluate_agent(self, num_episodes=100, max_steps=200):
        print(f"\n=== Evaluating Random Agent on {self.env_name} for {num_episodes} episodes ===")
        rewards = []
        for _ in range(num_episodes):
            episode_reward = self.run_episode(max_steps=max_steps)
            rewards.append(episode_reward)

        # 9. Calculate and display the total reward and average reward
        avg_reward = np.mean(rewards)
        print(f"\nAverage reward over {num_episodes} episodes: {avg_reward:.2f}")
        self.env.close()
        return avg_reward
```

## 2. Test on CartPole Environment

Let's apply the `RandomAgent` to the `CartPole-v1` environment.

```python
# Create and evaluate the RandomAgent for CartPole
cartpole_agent = RandomAgent('CartPole-v1')
cartpole_agent.evaluate_agent(num_episodes=10, max_steps=500) # CartPole-v1 has a default max_episode_steps of 500
```

## 3. Test on MountainCar Environment

Next, let's apply the `RandomAgent` to the `MountainCar-v0` environment.

```python
# Create and evaluate the RandomAgent for MountainCar
mountaincar_agent = RandomAgent('MountainCar-v0')
mountaincar_agent.evaluate_agent(num_episodes=100, max_steps=200) # MountainCar-v0 has a default max_episode_steps of 200
```

## 4. Test on FrozenLake Environment

Finally, let's apply the `RandomAgent` to the `FrozenLake-v1` environment.

```python
# Adjusting for Gymnasium: to pass render_mode or other args, typically we do it in gym.make
# The current RandomAgent class does not support passing env creation args like is_slippery=False directly.
# Let's modify the RandomAgent to allow for this.

# Re-instantiate the RandomAgent with a slightly modified constructor to handle custom env kwargs.
class RandomAgentModified(RandomAgent):
    def __init__(self, env_name, render_mode=None, **kwargs):
        # For FrozenLake, we might want to specify is_slippery=False or map_name
        self.env = gym.make(env_name, render_mode=render_mode, **kwargs)
        self.env_name = env_name
        print(f"Environment '{env_name}' created successfully with args: {kwargs}")

# Now create and evaluate the RandomAgentModified for FrozenLake
frozenlake_agent = RandomAgentModified('FrozenLake-v1', is_slippery=False, map_name="4x4")
frozenlake_agent.evaluate_agent(num_episodes=1000, max_steps=100)
```
